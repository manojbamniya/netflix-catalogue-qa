```python
def describe_and_combine(row):
    """
    Combines description, genres, duration, cast, and director into one string.
    Adds labels (Genres:, Duration:, etc.) and handles both text strings and lists.
    Skips missing data, empty strings, or empty lists completely.
    """
    parts = []
    
    # 1. Add description if it exists
    if pd.notna(row['description']) and str(row['description']).strip():
        parts.append(str(row['description']).strip())
        
    # 2. Add columns with their labels sequentially
    columns_to_check = [
        ("listed_in", "Genres"), 
        ("duration", "Duration"),
        ("cast", "Cast"), 
        ("director", "Director")
    ]
    
    for column, label in columns_to_check:
        val = row[column]
        
        if pd.notna(val):
            # If it's a list (and has items), join them with commas
            if isinstance(val, list) and len(val) > 0:
                list_as_str = ", ".join(map(str, val))
                parts.append(f"{label}: {list_as_str}.")
            
            # If it's a regular string (and not empty spaces), append it
            elif isinstance(val, str) and val.strip():
                parts.append(f"{label}: {val.strip()}.")
                
    # Combine everything with spaces
    combined_text = " ".join(parts)
    
    # Optional Cleanup: Remove any double spaces or isolated trailing commas from lists
    cleaned_words = [word.strip(',') for word in combined_text.split() if word.strip(',') != '']
    
    return " ".join(cleaned_words)

# Apply it to your DataFrame (make sure your variable name matches 'movies' or 'df')
movies['movie_description'] = movies.apply(describe_and_combine, axis=1)
```
